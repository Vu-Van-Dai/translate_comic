```python

import torch
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image
from google.colab import files

# 1. Lấy ảnh gốc và mask đã mở rộng
original = np.array(raw.convert("RGB"))
mask = np.array(
    Image.open("/content/mask_expanded.png").convert("L")
)

h, w = original.shape[:2]
assert mask.shape == (h, w), "Sai kích thước mask!"

selected = mask >= 128

# 2. Padding lên bội số của 8
ph, pw = (-h) % 8, (-w) % 8

padded_image = np.pad(
    original,
    ((0, ph), (0, pw), (0, 0)),
    mode="edge"
)

padded_mask = np.pad(
    selected.astype(np.float32),
    ((0, ph), (0, pw)),
    mode="constant"
)

image_tensor = (
    torch.from_numpy(padded_image.copy())
    .permute(2, 0, 1)
    .unsqueeze(0)
    .to("cuda", dtype=torch.float32) / 255.0
)

mask_tensor = (
    torch.from_numpy(padded_mask.copy())
    .unsqueeze(0).unsqueeze(0)
    .to("cuda", dtype=torch.float32)
)

# 3. Chạy model Anime/Manga LaMa
model.eval()

with torch.inference_mode():
    result = model(image_tensor, mask_tensor)
    torch.cuda.synchronize()

cleaned = (
    result[0, :, :h, :w]
    .permute(1, 2, 0)
    .clamp(0, 1)
    .cpu().numpy() * 255
).round().astype(np.uint8)

# Giữ nguyên các pixel ngoài mask
cleaned[~selected] = original[~selected]

# 4. Lưu ảnh
output_path = "/content/cleaned_expanded.png"
Image.fromarray(cleaned).save(output_path)

# 5. So sánh với kết quả trước
old_path = "/content/cleaned_manual.png"

fig, axes = plt.subplots(1, 3, figsize=(18, 9))

axes[0].imshow(original)
axes[0].set_title("Ảnh gốc")

if __import__("os").path.exists(old_path):
    axes[1].imshow(Image.open(old_path))
    axes[1].set_title("Clean lần 1")
else:
    axes[1].imshow(mask, cmap="gray")
    axes[1].set_title("Mask")

axes[2].imshow(cleaned)
axes[2].set_title("Clean lần 2 - Mask +4px")

for ax in axes:
    ax.axis("off")

plt.tight_layout()
plt.show()

print("✅ Hoàn thành!")
files.download(output_path)
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image

# Đọc mask đã vẽ trên Gradio
mask = np.array(
    Image.open("/content/mask_expanded.png").convert("L")
)

# Mở rộng mask thêm 4 pixel theo mọi hướng
kernel = cv2.getStructuringElement(
    cv2.MORPH_ELLIPSE,
    (9, 9)
)

expanded_mask = cv2.dilate(
    mask,
    kernel,
    iterations=1
)

Image.fromarray(expanded_mask).save(
    "/content/mask_expanded.png"
)

# Hiển thị so sánh
fig, axes = plt.subplots(1, 2, figsize=(12, 6))

axes[0].imshow(mask, cmap="gray")
axes[0].set_title("Mask ban đầu")

axes[1].imshow(expanded_mask, cmap="gray")
axes[1].set_title("Mask mở rộng 4px")

for ax in axes:
    ax.axis("off")

plt.show()
```

```python

import io
import re
import secrets
import threading

import numpy as np
import torch
import uvicorn

from PIL import Image
from fastapi import FastAPI, UploadFile, File, Form, Header, HTTPException
from fastapi.responses import Response

app = FastAPI(title="Comic GPU API")

# Khóa bảo vệ API
API_KEY = secrets.token_urlsafe(32)

# Không cho nhiều request dùng GPU cùng lúc
gpu_lock = threading.Lock()

# Model LaMa đã được nạp trên GPU T4
model.eval()

@app.get("/health")
def health():
    return {
        "status": "online",
        "gpu": torch.cuda.get_device_name(0),
        "model": "Anime Manga LaMa",
        "device": "cuda:0"
    }

@app.post("/inpaint")
def inpaint(
    page: str = Form(...),
    image: UploadFile = File(...),
    mask: UploadFile = File(...),
    x_api_key: str = Header(default="", alias="X-API-Key")
):
    # Kiểm tra API Key
    if not secrets.compare_digest(x_api_key, API_KEY):
        raise HTTPException(401, "Invalid API key")

    # Kiểm tra số trang
    if not re.fullmatch(r"\d{3,4}", page):
        raise HTTPException(400, "Page phải là 001, 004, 012...")

    try:
        image_bytes = image.file.read(15_000_001)
        mask_bytes = mask.file.read(15_000_001)

        if len(image_bytes) > 15_000_000 or len(mask_bytes) > 15_000_000:
            raise HTTPException(413, "File vượt giới hạn 15MB")

        original_image = Image.open(
            io.BytesIO(image_bytes)
        ).convert("RGB")

        mask_image = Image.open(
            io.BytesIO(mask_bytes)
        ).convert("L")

        if original_image.size != mask_image.size:
            raise HTTPException(400, "Ảnh và mask khác kích thước")

        width, height = original_image.size

        if width * height > 4_000_000:
            raise HTTPException(413, "Ảnh vượt giới hạn 4MP")

        original = np.array(original_image)
        selected = np.array(mask_image) >= 128

        if not selected.any():
            raise HTTPException(400, "Mask không có vùng trắng")

        # Padding lên bội số của 8
        ph = (-height) % 8
        pw = (-width) % 8

        padded_image = np.pad(
            original,
            ((0, ph), (0, pw), (0, 0)),
            mode="edge"
        )

        padded_mask = np.pad(
            selected.astype(np.float32),
            ((0, ph), (0, pw)),
            mode="constant"
        )

        image_tensor = (
            torch.from_numpy(padded_image.copy())
            .permute(2, 0, 1)
            .unsqueeze(0)
            .to("cuda:0", dtype=torch.float32) / 255.0
        )

        mask_tensor = (
            torch.from_numpy(padded_mask.copy())
            .unsqueeze(0).unsqueeze(0)
            .to("cuda:0", dtype=torch.float32)
        )

        # Chạy LaMa trên T4
        with gpu_lock, torch.inference_mode():
            result = model(image_tensor, mask_tensor)
            torch.cuda.synchronize()

        cleaned = (
            result[0, :, :height, :width]
            .permute(1, 2, 0)
            .clamp(0, 1)
            .cpu().numpy() * 255
        ).round().astype(np.uint8)

        # Giữ nguyên mọi pixel ngoài mask
        cleaned[~selected] = original[~selected]

        buffer = io.BytesIO()
        Image.fromarray(cleaned).save(buffer, format="PNG")

        return Response(
            content=buffer.getvalue(),
            media_type="image/png",
            headers={
                "Content-Disposition":
                    f'attachment; filename="{page}_clean_v01.png"'
            }
        )

    except HTTPException:
        raise
    except (OSError, ValueError, Image.DecompressionBombError) as e:
        raise HTTPException(400, f"Ảnh không hợp lệ: {e}")

# Khởi chạy server trong background thread
server = uvicorn.Server(
    uvicorn.Config(
        app,
        host="127.0.0.1",
        port=8000,
        log_level="warning"
    )
)

threading.Thread(
    target=server.run,
    daemon=True
).start()

print("✅ FastAPI đang khởi động trên cổng 8000")
print("API_KEY:", API_KEY)
```
